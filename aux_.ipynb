```python
import os
from tqdm import tqdm
from edgar import Company
from edgar import set_identity
import pandas as pd

import src.config as config

name = os.getenv("NAME", "Your Name")
email = os.getenv("EMAIL", "Your Email")

set_identity(f"{name} {email}")


sp500 = pd.read_csv(os.path.join(config.general['paths']['input'], 's&p500.csv'), sep=';', dtype={'CIK': str})
pbar = tqdm(zip(sp500.CIK, sp500.Symbol), total=len(sp500), desc="Processing CIKs")
symbols, filings = [], []
for cik, symbol in pbar:
    element = symbol if pd.isna(cik) else cik
    pbar.set_description(f"Processing CIK: {element}")
    company = Company(element)
    symbols.append(element)
    filings.append(len(company.get_filings()))

df = pd.DataFrame({"Symbol": symbols, "Filings": filings}).sort_values(by="Filings", ascending=True)
```
